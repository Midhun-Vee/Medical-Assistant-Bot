```python
import pandas as pd
import os
```

```python

CSV_DIR = "datasets/csv_datasets/raw_datasets"

PATIENT_DOCTOR_FILE = os.path.join(
    CSV_DIR,
    "Healthcare Patient-Doctor Conversation Dataset (50K Samples).csv",
)

MEDQUAD_FILE = os.path.join(
    CSV_DIR,
    "medquad.csv",
)

MTSAMPLES_FILE = os.path.join(
    CSV_DIR,
    "mtsamples.csv",
)

OUTPUT_PATH = "datasets/csv_datasets/preprocessed_datasets"
```

```python
data1 = pd.read_csv(PATIENT_DOCTOR_FILE)
data2 = pd.read_csv(MEDQUAD_FILE)
data3 = pd.read_csv(MTSAMPLES_FILE)
```
